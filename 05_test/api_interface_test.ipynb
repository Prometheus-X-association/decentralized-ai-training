## Registering a deployment

```python
ORCHESTRATOR_ENDPOINT = "http://orchestrator:8000"
ORCHESTRATOR_CREDENTIALS = {
    "username": "test",
    "password": "orch_test"
}
MLFLOW_ENDPOINT = "http://aggregator:5000"
MLFLOW_CREDENTIALS = {
    "username": "test",
    "password": "mlflow_test"
}
```

```python
data = dict()
data["orchestrator_uri"] = ORCHESTRATOR_ENDPOINT
data["orchestrator_credentials"] = ORCHESTRATOR_CREDENTIALS
data["mlflow_uri"] = MLFLOW_ENDPOINT
data["mlflow_credentials"] = MLFLOW_CREDENTIALS

payload = json.dumps(data)
```

```python
response = requests.put(
    url = f"{API_URI}/register_deployment",
    params = {"ptx_contract_id": "blabla"},
    json = data
)

response.json()
```

## Check deployment

```python
response = requests.get(f"{API_URI}/get_federated_mlflow_access",
                        params={"ptx_contract_id": "blabla"})
print(response.json())
```

```python
response = requests.post(f"{API_URI}/start_training",
                         params={"ptx_contract_id": "blabla"},
                         json={"training_architecture": {}})
response.json()
```

## Remove deployment

```python
response = requests.delete(f"{API_URI}/remove_deployment",
                           params={"ptx_contract_id": "blabla"})

response.json()
```
