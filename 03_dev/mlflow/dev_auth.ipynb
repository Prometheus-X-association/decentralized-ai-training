## Adding users to MLFlow with different permissions

| user name | role | MLFLow permission |
|-----------|------|-------------------|
| aggregator | saves models (in cluster) | MANAGE |
| model_user | downloads models (outside cluster) | USE |

```python
MLFLOW_TRACKING_URI = "http://localhost:5555"
```

```python
# login to mlflow server as admin:

import mlflow
from mlflow.server import get_app_client

import os

os.environ["MLFLOW_TRACKING_USERNAME"] = "admin"
os.environ["MLFLOW_TRACKING_PASSWORD"] = "password1234"
os.environ["MLFLOW_TRACKING_URI"] = MLFLOW_TRACKING_URI

client = get_app_client("basic-auth", MLFLOW_TRACKING_URI)
```

```python
#creating users:

client.create_user(username="aggregator", password="password1234")
client.create_user(username="model_user", password="password1234")
```

```python
mlflow.set_tracking_uri(MLFLOW_TRACKING_URI)
experiment_id = mlflow.create_experiment("training")
```
