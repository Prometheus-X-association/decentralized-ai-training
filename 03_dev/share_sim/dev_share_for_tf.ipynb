```python
import tensorflow as tf
from tensorflow import keras
```

```python
model = keras.Sequential(
    [
        keras.Input(shape=(3,)),
        keras.layers.Dense(32, activation="relu"),
        keras.layers.Dense(16, activation="relu"),
        keras.layers.Dense(1, activation="sigmoid")
    ]
)
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
import json

model_dict = json.loads(model.to_json())
with open("model.json", "w") as f:
    json.dump(model_dict, f)
```

```python
with open("model.json", "r") as f:
    model_json = json.load(f)
model2 = keras.models.model_from_json(json.dumps(model_json))
```

```python
import numpy as np

for layer in range(len(model.layers)):
    weights = model.layers[layer].get_weights()
    weight_noise_w_1 = tf.random.uniform(shape=weights[0].shape, minval=-0.2, maxval=0.2)
    weight_noise_b_1 = tf.random.uniform(shape=weights[1].shape, minval=-0.2, maxval=0.2)
    weight_noise_w_2 = -1*weight_noise_w_1
    weight_noise_b_2 = -1*weight_noise_b_1
    model.layers[layer].set_weights([weights[0] + weight_noise_w_1, weights[1] + weight_noise_b_1])
```
